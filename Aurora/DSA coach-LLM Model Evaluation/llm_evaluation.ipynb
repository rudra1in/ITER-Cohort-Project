```python
import time

questions = [
    {
        "question": "What is the time complexity of accessing an array element by index?",
        "expected": "O(1)"
    },
    {
        "question": "What is the difference between a stack and a queue?",
        "expected": "Stack follows LIFO, while queue follows FIFO."
    },
    {
        "question": "What is binary search and what is its time complexity?",
        "expected": "Binary search repeatedly divides the sorted search space in half. Time complexity is O(log n)."
    },
    {
        "question": "What data structure is commonly used for BFS?",
        "expected": "Queue"
    },
    {
        "question": "What is the average lookup complexity of a hash table?",
        "expected": "O(1)"
    },
    {
        "question": "Explain Two Sum.",
        "expected": "Use a hash map to find two numbers whose sum equals the target, with O(n) average time complexity."
    },
    {
        "question": "What is a linked list?",
        "expected": "A linked list consists of nodes where each node stores data and a reference to the next node."
    },
    {
        "question": "What is the time complexity of searching in a linked list?",
        "expected": "O(n)"
    }
]
```

```python
results = []

for item in questions:

    question = item["question"]

    print("\n" + "=" * 70)
    print("QUESTION:", question)
    print("=" * 70)

    for name, llm in llm_models.items():

        start_time = time.perf_counter()

        response = llm.invoke(question)

        end_time = time.perf_counter()

        latency = end_time - start_time

        # Clean response
        if isinstance(response.content, str):
            answer = response.content

        elif isinstance(response.content, list):

            parts = []

            for part in response.content:

                if isinstance(part, dict) and "text" in part:
                    parts.append(part["text"])

                elif isinstance(part, str):
                    parts.append(part)

            answer = "\n".join(parts)

        else:
            answer = str(response.content)

        results.append({
            "model": name,
            "question": question,
            "answer": answer,
            "latency": latency
        })

        print(f"\nMODEL: {name}")
        print(f"Latency: {latency:.2f} seconds")
        print("Answer:")
        print(answer)
```
